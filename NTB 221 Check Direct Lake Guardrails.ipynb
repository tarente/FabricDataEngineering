```python
from pyspark.sql.functions import *
```

```python
lakehouse       = "Lakehouse Name containing the shortcuts" # The Fabric lakehouse name or ID. Defaults to None which resolves to the lakehouse attached to the notebook.
lakehouseSchema = "Schema enabled Lakehouse Name"           # The Fabric lakehouse name or ID. Defaults to None which resolves to the lakehouse attached to the notebook.
workspace       = "<Workspace Name>"               # The Fabric workspace name or ID used by the lakehouse. Defaults to None which resolves to the workspace of
                                              # the attached lakehouse or if no lakehouse attached, resolves to the workspace of the notebook.
extended        = True                        # Obtains additional columns relevant to the size of each table.
count_rows      = True                        # Obtains a row count for each lakehouse table.
export          = False                       # Exports the resulting dataframe to a delta table in the lakehouse.
```

```python
dfCheckGuardRailsSchema = (
    spark
        .createDataFrame (
            get_lakehouse_tables(
                lakehouse = lakehouseSchema,
                workspace = workspace,
                extended = extended,
                count_rows = count_rows,
                export = export
            )
        )
        .withColumn("Table Name", concat ("Schema Name", lit("."), "Table Name"))
        .withColumn("Schema Name", lit("dbo"))
        .select (
            "Workspace Name",
            "Lakehouse Name",
            "Schema Name",
            "Table Name",
            "Format",
            "Type",
            "Files",
            "Row Groups",
            "Table Size",
            "Row Count",
            "SKU"
        )
)
```

```python
dfCheckGuardRails = (
    spark
        .createDataFrame (
            get_lakehouse_tables(
                lakehouse = lakehouse,
                workspace = workspace,
                extended = extended,
                count_rows = count_rows,
                export = export
            )
        )
        .withColumn("Lakehouse Name", lit(lakehouseSchema))
        .select (
            "Workspace Name",
            "Lakehouse Name",
            "Schema Name",
            "Table Name",
            "Format",
            "Type",
            "Files",
            "Row Groups",
            "Table Size",
            "Row Count",
            "SKU"
        )
)
```
